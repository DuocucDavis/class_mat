```python
#Método 1: Fórmula general#
a1 = 5     # primer término
d = 3      # diferencia
n = 10     # término que queremos

an = a1 + (n - 1) * d
print(f"Término {n} =", an)
```

```python
#Método 2: Usar bucle for
a1, d = 5, 3
n = 10
sucesion = [a1 + i * d for i in range(n)]
print(sucesion)
```

```python
#2. Sucesión Geométrica
#Método 1: Fórmula general
a1, r, n = 2, 3, 8
an = a1 * (r ** (n - 1))
print(f"Término {n} =", an)



```

```python
#Método 2: Bucle con multiplicación
a1, r = 2, 3
n = 8
sucesion = [a1 * (r ** i) for i in range(n)]
print(sucesion)
```

```python
#3. Sucesión Recurrente
#Método 1: Iterativo
n = 10
a = [0, 1]
for i in range(2, n):
    a.append(a[i-1] + a[i-2])
print(a)
```

```python
#Método 2: Recursivo
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

n = 10
sucesion = [fibonacci(i) for i in range(n)]
print(sucesion)
```

```python
#Sucesión Definida por Fórmula General

def formula(n):
    return n**2 + 2*n + 1

sucesion = [formula(i) for i in range(1, 11)]
print(sucesion)
```

```python
#Sucesión Mixta o Personalizada
sucesion = [(n**2 if n % 2 == 0 else n + 5) for n in range(1, 11)]
print(sucesion)
```

```python
#Sucesión con Numpy (eficiente)
import numpy as np

a1, d = 3, 2
n = 10
n_values = np.arange(n)
sucesion = a1 + d * n_values
print(sucesion)
```

```python
#Graficar una sucesión
import matplotlib.pyplot as plt

a1, d = 3, 2
n = 10
x = range(1, n+1)
y = [a1 + (i-1)*d for i in x]

plt.plot(x, y, marker='o')
plt.title("Sucesión Aritmética")
plt.xlabel("n")
plt.ylabel("a_n")
plt.grid(True)
plt.show()
```

```python
#Sucesiones con Condiciones Avanzadas
def sucesion(n):
    return 2*n if n < 5 else n**2

serie = [sucesion(i) for i in range(1, 11)]
print(serie)
```

```python
#exponenciación de matrices (Fibonacci)
def mat_mult(A, B):
    return [[A[0][0]*B[0][0] + A[0][1]*B[1][0],
             A[0][0]*B[0][1] + A[0][1]*B[1][1]],
            [A[1][0]*B[0][0] + A[1][1]*B[1][0],
             A[1][0]*B[0][1] + A[1][1]*B[1][1]]]

def mat_pow(M, n):
    # exponenciación binaria
    result = [[1,0],[0,1]]  # identidad
    while n > 0:
        if n & 1:
            result = mat_mult(result, M)
        M = mat_mult(M, M)
        n >>= 1
    return result

def fib_matrix(n):
    if n == 0: return 0
    M = [[1,1],[1,0]]
    P = mat_pow(M, n-1)
    return P[0][0]  # F_n
```
